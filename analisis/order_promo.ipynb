Building an End-to-End Data Engineering Pipeline for E-Commerce Order Analytics

Step 1: EXTRACT - Baca Data Mentah

```python
from pathlib import Path
import pandas as pd
import numpy as np  
import ast
# Import semua data raw 
data = pd.read_csv("../data/raw/raw_order_promo.csv")

data.head(10)
```

```python
# Inspeksi awal
print(f"Jumlah baris: {len(data)}")
print(f"Kolom: {list(data.columns)}")
data.info()
```

```python
print("\nDuplikasi")
print(f"{data.duplicated().sum()}")
```

```python
# Mencari data duplikasi
duplikat = data[data.duplicated(keep=False)]
print(duplikat)
```

```python
print("\nMissing values:")
print(data.isnull().sum())
```

Step 2: TRANSFORM - Bersihkan Data

```python
# Menghapus data duplicate karena memang sama kode promonya
data = data.drop_duplicates()
```

```python
# Mengubah kode promo menjadi hurup besar semua atau UPPER
# Mengubah promo_code menjadi huruf besar semua
data['promo_code'] = (data['promo_code'].astype(str).str.strip().str.upper())
# Menampilkan hasil
print(data[['promo_code']])
```

```python
data.info
```

```python
# Simpan ke CSV data setelah proses atau clean
data.to_csv(
    "../data/warehouse/order_promo_clean.csv",
    index=False,
    encoding="utf-8"
)
# Melihat kembali data yang disimpan
data = pd.read_csv("../data/warehouse/order_promo_clean.csv")
data.head(10)
```
