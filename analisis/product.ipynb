```python
from pathlib import Path
import pandas as pd
import numpy as np  
import ast
# Import semua data raw 
data = pd.read_csv("../data/raw/raw_products.csv")

data.head(10)
```

```python
data.sort_values(
    by='harga_satuan',
    ascending=False
).head(10)
```

Visualisasi Perbandingan Katergori Produk

```python
import matplotlib.pyplot as plt

# Hitung jumlah produk per kategori
kategori = data['kategori'].value_counts()
plt.figure(figsize=(6,4))
kategori.plot(kind='bar')
plt.title('Jumlah Produk Berdasarkan Kategori')
plt.xlabel('Kategori')
plt.ylabel('Jumlah Produk')
plt.xticks(rotation=0)
plt.show()
```
